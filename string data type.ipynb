```python
#Q-1 Write a Python program to count a particular character in a given string using while loop.
data="Python Programming"
strcount=0
i=0
while (i<len(data)):
    if data[i]=='P':
        strcount+=1
    i+=1
print(strcount)
```

```python
#Q-2 Write a python program to remove all vowels of any string using for loop
data="Python Programming"
result=""
for i in data:
    if i not in "aeiouAEIOU":
        result = result + i
print(result)
    
```

```python
#Q-3 remove @ from the email.
##Using for loop
data="invertis@gmail.com"
result=""
for i in data:
    if i not in "@": #using membership operator
        result+=i
print(result)
    
```

```python
#Using For loop
data="invertis@gmail.com"
result=""
for i in data:
    if i != "@": #using assignment operator
        result+=i
print(result)
```

```python
# using for while loop
data="invertis@gmail.com"
result=""
i=0
while i < len(data):
    if data[i] != "@":
        result+= data[i]
    i+=1
print(result)
    
```
